```python
import pandas as pd
import numpy as np

df = pd.read_csv('clinic_intake.csv')
df
```

```python
df.isnull().sum()
```

```python
df["age"] = df['age'].fillna(df['age'].mean())
df["bmi"] = df['bmi'].fillna(df['bmi'].mean())
df = df.dropna()
df.isnull().sum()
```

```python
df
```

```python
df = df.drop(columns=['patient_id'])
df
```

```python
df.shape
```

```python
df = pd.get_dummies(df, columns=['sex', 'region', 'smoker'])
df
```

```python
df.shape
```

```python
amin, amax = df['age'].min(), df['age'].max()
df['age'] = (df['age'] - amin) / (amax - amin)

bmin, bmax = df['bmi'].min(), df['bmi'].max()
df['bmi'] = (df['bmi'] - bmin) / (bmax - bmin)

df
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test = train_test_split(df, test_size= 0.2, random_state=1)
print(x_train.shape, x_test.shape)
```
